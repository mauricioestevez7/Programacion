## 1. Argumentos y referencias compartidas
### Paso de objetos inmutables:

```python
def f(a):	    	# a is assigned to (references) the passed object
	a = 99 			# Changes local variable a only

b = 88		
f(b)				# a and b both reference same 88 initially
print("b = ", b)	# b is not changed
```

### Paso de objetos mutables:

```python
def changer(a, b):	    # Arguments assigned references to objects
		a = 2			# Changes local name's value only
		b[0] = 'spam' 	# Changes shared object in place
	
					
x = 1
l = [1, 2]
changer(x, l)           # Caller: Pass immutable and mutable objects
print("l = ", l)        # ['spam', 2]  # x is unchanged, l is different!
assert x == 1, "x es 1"
```

### Cómo evitar que una rutina modifique los elementos mutables:
#### - Pasar una copia

```python
l = [1, 2]
changer(x, l[:]) 	# Pass a copy, so our 'l' does not change 

print("l = ", l) 
```

#### - Utilizar invariantes (en las estructuras de datos de entrada)

```python
l = [1, 2]
def changer(a, b):
    b = b[:]		# Copy input list so we don't impact caller
    a = 2
    b[0] = 'spam'	# Changes our list copy only

changer(x, l)
assert l == [1, 2], "l es una invariante"
print("l = ", l) 
```

#### - Convertir el objeto mutable a un objeto inmutable:

```python
l = [1, 2]
def changer(a, b):	   
		a = 2			
		b[0] = 'spam'   # TypeError: 'tuple' object does not support item assignment

changer(x, tuple(l))	# Pass a tuple, so changes are errors

print("l = ", l)
assert l == [1, 2], "l es una invariante"
```

## 2 modos especiales de correspondencia de argumentos
#### ver figura 06 en Drive
###### Ten cuidado de no confundir la sintaxis especial «nombre=valor» en la cabecera de una función y en una llamada a una función; en la llamada, representa un argumento por nombre (argumento de palabra clave), mientras que en la cabecera especifica un valor predeterminado para un argumento opcional.

### Ejemplos de palabras clave y valores predeterminados
#### 1. Comportamiento por defecto:

```python
def f(a, b, c): 
    print(a, b, c)

f(1, 2, 3)          # 1 2 3
```

#### 2. Palabras clave
###### El orden de los argumentos de izquierda a derecha ya no importa cuando se utilizan palabras clave, ya que los argumentos se identifican por su nombre, no por su posición.

```python
def f(a, b, c): 
    print(a, b, c)func(*(1, 2), **{'d': 4, 'c': 3})       # Same as func(1, 2, d=4, c=3)
# 1 2 3 4
func(1, *(2, 3), **{'d': 4})            # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, c=3, *(2,), **{'d': 4})         # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
func(1, *(2, 3), d=4)                   # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, *(2,), c=3, **{'d':4})      # Same as func(1, 2, c=3, d=4)
# 1 2 3 4

f(c=3, b=2, a=1)    # 1 2 3
```

```python
f(1, c=3, b=2)		# a gets 1 by position, b and c passed by name
                    # 1 2 3

# they make your calls a bit more self documenting
# f(name='Bob', age=40, job='dev')
```

#### 3. Incumplimientos
###### Los valores predeterminados nos permiten hacer opcionales determinados argumentos de una función; si no se les asigna un valor, el argumento recibe su valor predeterminado antes de que se ejecute la función.

```python
def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)


f(1)        # 1 2 3
f(a=1)      # 1 2 3
```

###### Si pasamos dos valores, solo c toma su valor predeterminado; y con tres valores, no se utiliza ningún valor predeterminado:

```python
def f(a, b=2, c=3): 	# a required, b and c optional
	print(a, b, c)
    
f(1, 4)     # 1 4 3
f(1, 4, 5)	# Override defaults: 1 4 5

f(1, c=6)	# Choose defaults: 1 2 6
```

#### 4. Combinación de palabras clave y valores predeterminados
###### Observa, una vez más, que cuando se utilizan argumentos por palabra clave en la llamada, el orden en el que aparecen los argumentos no importa; Python los hace corresponder por nombre, no por posición.

###### Quien realiza la llamada debe proporcionar valores para «spam» y «eggs», pero estos pueden asignarse por posición o por nombre.

```python
def func(spam, eggs, toast=0, ham=0):
	print((spam, eggs, toast, ham)) 	# First 2 required

func(1, 2)							# Output: (1, 2, 0, 0)
func(1, ham=1, eggs=0)				# Output: (1, 0, 0, 1)
func(spam=1, eggs=0)				# Output: (1, 0, 0, 0)
func(toast=1, eggs=2, spam=3)		# Output: (3, 2, 1, 0)
func(1, 2, 3, 4) 					# Output: (1, 2, 3, 4)
```

#### 5. Argumentos de solo palabra clave en Python 3.X
###### También podemos utilizar un carácter `*` por sí solo en la lista de argumentos para indicar que una función no admite una lista de argumentos de longitud variable, pero que, aun así, exige que todos los argumentos que siguen al `*` se pasen como argumentos por palabra clave.

```python
def kwonly(a, *b, c):
    print(a, b, c)

kwonly(1, 2, c=3)       # 1 (2,) 3

kwonly(a=1, c=3)        # 1 () 3

kwonly(1, 2, 3)         # TypeError: kwonly() missing 1 required keyword-only argument: 'c'
```

```python
def kwonly(a, *, b='spam', c='ham'):
	print(a, b, c)

kwonly(1)               # 1 spam ham
kwonly(1, c=3)          # 1 spam 3
kwonly(a=1)             # 1 spam ham
kwonly(c=3, b=2, a=1)   # 1 2 3
kwonly(1, 2)            # TypeError: kwonly() takes 1 positional argument but 2 were given
```

```python
def kwonly(a, *, b, c='spam'):
    print(a, b, c)

kwonly(1, c='eggs')     # TypeError: kwonly() missing 1 required keyword-only argument: 'b'
```

### Ejemplos de argumentos arbitrarios
###### Las dos últimas extensiones de correspondencia, * y **, están diseñadas para admitir funciones que aceptan cualquier número de argumentos.

###### Ambas pueden aparecer tanto en la definición de una función como en una llamada a una función.

#### Headers: Recopilación de argumentos
###### En la definición de la función, agrupa los argumentos posicionales no emparejados en una tupla y asigna dicha tupla a la variable `args`.

```python
def f(*args):
    print(args)

f()                 # ()
f(1,)               # (1,)
f(1, 2, 3, 4)       # (1, 2, 3, 4)  
```

###### La funcionalidad ** solo funciona con argumentos por palabra clave. Los agrupa en un nuevo diccionario.

```python
def f(**args):
    print(args)
f()                 # {}
f(a=1, b=2)         # {'a': 1, 'b': 2}
```

```python
def f(a, *pargs, **kargs):
    print(a, pargs, kargs)
f(1, 2, 3, x=1, y=2)        # 1 (2, 3) {'y': 2, 'x': 1}
```

#### Calls: Desglosando los argumentos
###### * En una llamada a una función, esta sintaxis desempaqueta una colección de argumentos.

```python
def func(a, b, c, d): 
    print(a, b, c, d)

args = (1, 2)
args += (3, 4)
func(*args)         # Same as func(1, 2, 3, 4)
```

```python
def func(a, b, c, d): 
    print(a, b, c, d)
args = {'a': 1, 'b': 2, 'c': 3}
args['d'] = 4
func(**args)    # Same as func(a=1, b=2, c=3, d=4)
func(*args)
```

```python
func(*(1, 2), **{'d': 4, 'c': 3})       # Same as func(1, 2, d=4, c=3)
# 1 2 3 4
func(1, *(2, 3), **{'d': 4})            # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, c=3, *(2,), **{'d': 4})         # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
func(1, *(2, 3), d=4)                   # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, *(2,), c=3, **{'d':4})      # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
```
